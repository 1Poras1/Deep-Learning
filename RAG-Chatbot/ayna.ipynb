```python
# Data Loading
from langchain_community.document_loaders import TextLoader

loader = TextLoader("Attention.txt", autodetect_encoding=True)
text_documents = loader.load()

print(text_documents[0].page_content[:1000])
```

```python
# Text Splitter
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=20
)

final_documents = text_splitter.split_documents(text_documents)

final_documents[0]
```

```python
# Create Embeddings
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
print(embeddings)
```

```python
#Embedding and vector store in vector_DB
from langchain_community.vectorstores import FAISS
db = FAISS.from_documents(final_documents, embeddings)
db
```

```python
query = input("Enter your query and question for Transformer: ")
docs = db.similarity_search(query)
for i in range(len(docs)):
    print(docs[i].page_content)
```
